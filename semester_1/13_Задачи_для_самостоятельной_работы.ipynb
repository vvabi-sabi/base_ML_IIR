# Лекция 13 — Алгоритмы и оценка сложности: задачи для самостоятельной работы

## Простые задачи


### 1. Сложность операций со списком и словарём
Создайте список и словарь на 1 000 000 элементов и замерьте (`time.perf_counter`) время операции `in` для каждого. Почему для словаря она занимает микросекунды, а для списка — миллисекунды?

```python
lst = [777] * 1_000_000
dct = {i: None for i in range(1_000_000)}
# замерьте время: 555 in lst и 555 in dct
```

Проверьте, что словарь как минимум в 10 раз быстрее.

**Подсказка:** поиск элемента в списке — O(n), в словаре — O(1) (хеш-таблица).


### 2. Подсчёт числа операций
Для функции ниже выведите формулу `f(n)` — число присваиваний как функцию от `n` — и проверьте её на нескольких значениях. Какова сложность функции?

```python
def f(n):
    a = n
    b = 2
    c = 0
    for i in range(n):
        c = c + a + b
    return c
```

**Подсказка:** посчитайте операции до цикла и операции одной итерации, умноженные на число итераций.


### 3. Линейный поиск
Реализуйте линейный поиск: верните индекс первого вхождения элемента в список или `-1`, если элемента нет. Добавьте счётчик числа сравнений и покажите, что в худшем случае их ровно `n`.

```python
data = [4, 2, 7, 1, 9, 3]
# linear_search(data, 7) -> 2
# linear_search(data, 100) -> -1
```

**Подсказка:** проходите `enumerate(data)` и сравнивайте каждый элемент с искомым.


### 4. Максимум списка
Реализуйте функцию `my_max(seq)` без встроенной `max()`: перебором за один проход. Проверьте на списке со случайными числами против `max()`.

**Подсказка:** запомните первый элемент и обновляйте «лучший», встречая больший.


### 5. Факториал итеративно
Реализуйте `fact(n)` — произведение чисел от 1 до n — циклом. Проверьте: `fact(0) == 1`, `fact(5) == 120`, `fact(10) == 3628800`. Какова сложность?

**Подсказка:** один цикл `for k in range(2, n + 1)` с накоплением произведения.


### 6. Простое число за O(√n)
Реализуйте `is_prime(n)`: переделите делители только до корня из n. Проверьте на всех числах от 2 до 50 против известного списка простых.

```python
# 1 и отрицательные — не простые; 2 — простое
```

**Подсказка:** если у `n` есть делитель больше корня, то есть и меньше корня: хватит условия `d * d <= n`.


### 7. Сумма цифр числа
Реализуйте `digit_sum(n)` без перевода в строку: остаток от деления на 10 и целочисленное деление. Проверьте: `digit_sum(9875) == 29`, `digit_sum(0) == 0`. Какова сложность?

**Подсказка:** цикл `while n > 0: s += n % 10; n //= 10`. Сложность равна числу цифр — O(log₁₀ n).


### 8. Переворот строки
Разверните строку, не используя срез `[::-1]`: двумя «указателями» с концов списка символов. Проверьте на `'abcd'`, `''` и палиндроме `'level'`.

**Подсказка:** меняйте местами символы на позициях `left` и `right`, двигая указатели навстречу.


### 9. Обмен значений без третьей переменной
Поменяйте значения переменных `a` и `b` местами дважды: сначала по схеме из лекции `k = a; a = b; b = k`, а затем без дополнительной переменной — через сложение/вычитание и через «кортежное» присваивание.

```python
a, b = 3, 7
# после обмена: a == 7, b == 3
```

**Подсказка:** `a = a + b; b = a - b; a = a - b`, а в Python достаточно `a, b = b, a`.


### 10. Очередь deque: вращение за O(1)
Реализуйте сдвиг списка влево на `k` позиций вручную (через `deque` и `popleft/append`), затем сравните с `list`-версией `lst[1:] + [lst[0]]` на 100 000 элементах. Насколько `deque.rotate(-1)` быстрее?

**Подсказка:** `deque` позволяет вставлять/удалять с обоих концов за O(1), а срез списка копирует все элементы — O(n).


## Средние задачи


### 11. Двоичный поиск за логарифм
Реализуйте двоичный поиск в отсортированном списке: верните индекс элемента (или `-1`) и число сравнений. Проверьте на отсортированном массиве из 1000 элементов, что сравнений не больше 11.

```python
a = list(range(0, 1000, 7))  # отсортирован
```

**Подсказка:** сравнение с серединой отрезка `[left, right]` и сужение отрезка вдвое; 2¹⁰ = 1024, поэтому 1024 элемента ищутся за ~10 сравнений.


### 12. Сортировка пузырьком с флагом
Реализуйте `bubble_sort(a)`, которая возвращает `(отсортированный список, число перестановок)`. Добавьте ранний выход, если за проход не было ни одной перестановки. Проверьте: `[5, 2, 9, 1] -> ([1, 2, 5, 9], 4)`, отсортированный список — 0 перестановок.

**Подсказка:** число перестановок пузырька равно числу инверсий в исходном списке.


### 13. Сортировка выбором и число обменов
Реализуйте `selection_sort(a)`: на каждом шаге ищите минимум хвоста и меняйте его на место. Верните `(список, число обменов)`. Докажите на случайных данных, что обменов никогда не больше `n - 1`.

```python
# [5, 2, 9, 1] -> ([1, 2, 5, 9], 2)
```

**Подсказка:** сравнений всегда O(n²), а обменов — не больше n−1, в этом преимущество пузырька... то есть выбора!


### 14. Сортировка вставками и число сдвигов
Реализуйте `insertion_sort(a)`, возвращающую `(список, число сдвигов)`. Проверьте: отсортированный список — 0 сдвигов, `[5, 4, 3, 2, 1]` — 10 сдвигов, `[5, 2, 9, 1]` — 4.

**Подсказка:** «выбранный элемент» вставляется в отсортированную левую часть, сдвигая большие элементы вправо; для отсортированных данных сложность O(n).


### 15. Баланс скобок (стек)
Реализуйте `is_balanced(s)` для скобок `()`, `[]`, `{}`: открывающие кладите в стек, закрывающие проверяйте с верхушкой стека. Проверьте: `'([]){}'` — True, `'([)]'` — False, `'(()'` — False, пустая строка — True.

**Подсказка:** строка сбалансирована, если в конце стек пуст и ни одна закрывающая скобка не «промахнулась».


### 16. Алгоритм Евклида: итерации и рекурсия
Реализуйте НОД двумя способами — итеративно (`a, b = b, a % b`) и рекурсивно, — возвращая число шагов. Проверьте: `gcd(462, 1071) == 21`, сверьтесь с `math.gcd` на всех парах до 40. Покажите «худший случай» — соседние числа Фибоначчи (55 и 34).

**Подсказка:** на каждой итерации пара `(a, b)` превращается в `(b, a mod b)`, пока `b` не станет нулём.


### 17. Фибоначчи: рекурсия против мемоизации
Реализуйте наивную рекурсивную `fib` со счётчиком вызовов: для `n = 10` должно получиться 177 вызовов, для `n = 20` — 21891. Затем оберните её в `functools.lru_cache` и покажите, что `fib(30) == 832040` считается мгновенно и за десятки вызовов.

**Подсказка:** число вызовов наивной реализации растёт как 2·F(n+1) − 1, то есть экспоненциально.


### 18. Анаграмма: O(n log n) и O(n)
Проверьте, является ли одна фраза анаграммой другой (регистр и пробелы игнорируются), двумя способами: через `sorted` и через `collections.Counter`. Проверьте на паре `'Astronomer'` / `'Moon starer'` и на неуспешном примере.

**Подсказка:** сортировка букв — O(n log n), подсчёт частот — O(n).


### 19. Two Sum: перебор и хеш-таблица
Дан список целых и целевая сумма: найдите индексы двух элементов с такой суммой. Реализуйте перебор O(n²) и решение за O(n) через словарь «значение → индекс». На случайных данных из 300 элементов сверьте, что оба решения находят пару (или её отсутствие).

```python
nums = [2, 7, 11, 15]  # target 9 -> (0, 1)
```

**Подсказка:** для каждого `x` ищите в словаре уже просмотренных число `target - x`.


### 20. Слияние двух отсортированных массивов
Реализуйте `merge(a, b)` — слияние двух отсортированных списков в один за O(n + m) двумя указателями без повторной сортировки. Сверьте с `sorted(a + b)` на случайных отсортированных данных.

**Подсказка:** на каждом шаге берите меньший из двух «головных» элементов; это ядро сортировки слиянием.


## Сложные задачи


### 21. Границы вхождения (left_bound / right_bound)
Реализуйте двоичный поиск левой и правой границы блока равных элементов: `left_bound(a, x)` — индекс первого элемента ≥ x, `right_bound(a, x)` — индекс первого > x. На `[1, 2, 2, 2, 2, 3, 5]` для `x = 2`: левая 1, правая 5 (четыре двойки). Сверьте с `bisect.bisect_left/right`.

**Подсказка:** используйте полуинтервал `[lo, hi)` и условие `a[mid] < x` (соответственно `<= x`).


### 22. Сортировка слиянием
Реализуйте `merge_sort(a)` рекурсией: деление пополам, сортировка половин, слияние. Проверьте на случайных массивах против `sorted()` (включая пустой). Сколько уровней рекурсии у массива из 1024 элементов?

**Подсказка:** T(n) = 2·T(n/2) + O(n) = O(n log n); глубина рекурсии log₂ n.


### 23. Быстрая сортировка с медианой из трёх
Реализуйте `quick_sort(a)` на месте: опора — медиана из первого, среднего и последнего элементов (защита от отсортированных данных), разделение по Ломуто. Проверьте против `sorted()` на случайных массивах.

**Подсказка:** медиана из трёх выбирается сравнением трёх значений; разделение — один проход «меньшие слева».


### 24. Решето Эратосфена
Реализуйте `sieve(n)` — все простые до n за O(n log log n): отмечайте кратные, начиная с d². Проверьте `sieve(20) == [2, 3, 5, 7, 11, 13, 17, 19]` и что простых до 100 ровно 25.

**Подсказка:** булев массив «не помечено» + цикл `for m in range(d * d, n + 1, d)`.


### 25. Лестница: число способов (мемоизация)
Подняться по лестнице из n ступеней можно на 1 или 2 ступеньки за шаг. Сколько способов добраться до вершины? Реализуйте `ways(n)` с `lru_cache`: `ways(3) == 3`, `ways(4) == 5`, `ways(10) == 89`. Найдите связь с числами Фибоначчи и биномиальными коэффициентами.

**Подсказка:** ways(n) = ways(n−1) + ways(n−2), ways(0) = ways(1) = 1; без мемоизации рекурсия сделала бы экспоненциально много повторных вызовов.


### 26. Ханойская башня
Реализуйте `hanoi(n)` — список ходов `(диск, откуда, куда)` для переноса башни со штыря A на C. Проверьте: `n = 3` даёт 7 ходов, и все ходы легальны (эмуляция с проверкой размеров). Сколько ходов для n = 10?

**Подсказка:** перенести n−1 на вспомогательный, положить n-й, перенести n−1 сверху; всего 2ⁿ − 1 — экспоненциальная сложность.


### 27. Отрезок с заданной суммой (два указателя)
В списке положительных чисел найдите отрезок с суммой, равной S, за O(n): два указателя и текущая сумма (скользящее окно). Верните `(начало, конец)` или `None`.

**Подсказка:** расширяйте окно вправо; если сумма превысила S — сдвигайте левый указатель.


### 28. Максимальная сумма подмассива (алгоритм Кадана)
Найдите максимальную сумму непрерывного подмассива за O(n) (в массиве есть отрицательные). Проверьте: `[-2, 1, -3, 4, -1, 2, 1, -5, 4] -> 6` (подмассив `[4, -1, 2, 1]`), в массиве одних отрицательных — максимум из них.

**Подсказка:** `current = max(x, current + x)` — либо продолжаем отрезок, либо начинаем заново.


### 29. Калькулятор обратной польской записи

### 30. Очередь из двух стеков
