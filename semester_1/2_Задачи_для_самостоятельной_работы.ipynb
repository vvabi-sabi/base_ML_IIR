# Задачи для самостоятельной работы

Лекция 2. **Списки, строки, циклы**

Решите каждую задачу в ячейке под её условием. Задачи упорядочены по сложности: **лёгкие (10)**, **средние (10)**, **сложные (10)**. Все темы соответствуют слайдам лекции (см. `themes_2.txt`).

Правила оформления: следуйте PEP8 (4 пробела на отступ, осмысленные имена переменных).

## Лёгкие задачи (10)

Строки и методы строк, списки, `range`, циклы, f-строки.

### 1. Склейка и дублирование строк

Даны два слова `first = "Python"` и `second = "Course"`. Склейте их в одну строку через `+`, продублируйте результат умножением на 2 и выведите через f-строку: полученную строку и её длину `len()`.

Темы: «Строки (str)», «Функции для работы со строками (сложение, дублирование)», «f-строки».

### 2. Изменяемые и неизменяемые типы

Даны строка `s = "Hello"` и список `l = [1, 2, 3]`.

а) Попробуйте изменить первый символ строки (`s[0] = "h"`) — запустите и объясните возникшую ошибку.

б) Измените первый элемент списка (`l[0] = 9`) и выведите `l`.

Темы: «Изменяемые и неизменяемые типы данных в Python», «Списки (list)».

### 3. Чётные числа через range

Выведите все чётные числа от 0 до 20 включительно, используя цикл `for` и `range(start, stop, step)` с шагом 2.

Темы: «Генератор арифметической прогрессии — range(start, stop, step)», «Цикл for».

### 4. Подсчёт буквы в строке

Дана строка `s = "banana"`. Циклом `for` по символам строки (итерация по строке) посчитайте, сколько раз в ней встречается буква `"a"`, и выведите результат.

Темы: «Строки (str)», «Итерация по строке (for el in A:)».

### 5. Методы регистра строки

Введите строку через `input()` и выведите её: в верхнем регистре (`.upper()`), в нижнем (`.lower()`), с заглавной первой буквой (`.capitalize()`).

Тема: «Методы строк (s.find, s.upper, s.split и т.д.)».

### 6. Добавление и удаление элементов списка

Дан список `a = [1, 2, 3]`. Добавьте в конец число 4 через `append`, добавьте число 0 в начало через `insert`, удалите элемент со значением 2 через `remove`, удалите последний элемент через `pop`. После каждой операции выведите список.

Темы: «Списки в Python (list)», «Добавление/удаление элемента из списка».

### 7. Поиск элемента в списке

Дан список `a = [10, 20, 30, 40]` и число `x = 30`. Проверьте двумя способами, есть ли `x` в списке: оператором `in` и циклом `for` с `if` (линейный поиск). Выведите результат обоих способов.

Темы: «Линейный поиск элемента в списке», «Итерация по списку».

### 8. Сумма элементов списка

Дан список чисел `nums = [5, 3, 8, 1]`. Найдите сумму всех его элементов циклом `for` (не используя встроенную `sum`), посчитайте среднее арифметическое и выведите оба значения.

Темы: «Списки в Python (list)», «Цикл for», «Итерация по списку».

### 9. Форматирование через f-строку

Даны переменные `name = "Аня"` и `age = 20`. Выведите с помощью одной f-строки сообщение: `Меня зовут Аня, мне 20 лет.` Затем удвойте возраст и выведите его внутри f-строки прямо в выражении `{age * 2}`.

Тема: «f-строки (форматирование)».

### 10. Разбиение строки на слова

Дано предложение `s = "Python — это здорово"`. Разбейте его на слова через `split()` и выведите: получившийся список и количество слов через `len()`.

Темы: «Методы строк (s.split)», «Строки (str)».

## Средние задачи (10)

Срезы, алгоритмы обращения и сдвига, поиск, `break`/`continue`/`else`, вложенные циклы.

### 11. Срезы строк и списков

Даны строка `s = "abcdefg"` и список `l = [1, 2, 3, 4, 5]`.

а) Выведите строку в обратном порядке срезом `[:: -1]` и список в обратном порядке.

б) Выведите каждый второй элемент списка с шагом 2 (`l[:: 2]`).

Темы: «Строки», «Списки», «Срезы» (см. примеры из `Lesson_2.ipynb`).

### 12. Обращение списка алгоритмом

Дан список `a = [2, 5, 4, 4, 5, 2]`. Обратите его **вручную** — циклом присвойте `a[i]` значение `a[n - 1 - i]` (алгоритм из лекции), без среза и без `reversed`.

Тема: «Алгоритм обращения списка».

### 13. Циклический сдвиг влево на одну позицию

Дан список `a = [1, 2, 3, 4, 5]`. Выполните циклический сдвиг влево на одну позицию: первый элемент становится последним, остальные сдвигаются влево (`[2, 3, 4, 5, 1]`). Реализуйте это вручную циклами.

Тема: «Алгоритм циклического сдвига в списке влево».

### 14. Максимум и его индекс

Дан список `a = [3, 7, 2, 9, 5]`. Найдите максимальный элемент и его индекс **циклом** (без `max` и `index`).

Темы: «Списки», «Итерация по списку», «Операторы сравнения».

### 15. Подсчёт положительных, отрицательных и нулей

Дан список чисел `nums = [1, -2, 0, 3, -4, 0, 5]`. Циклом посчитайте, сколько в нём положительных (`> 0`), отрицательных (`< 0`) и нулей (`== 0`). Выведите три числа.

Темы: «Операторы if и операторы сравнения», «Итерация по списку».

### 16. Лесенка из звёздочек

Выведите «лесенку» из 5 строк: на строке i (от 1 до 5) — звёздочка `"*"`, повторённая i раз. Используйте умножение строки на число и f-строку: `*`, `**`, `***` и так далее.

Темы: «Дублирование строк», «f-строки», «Цикл for».

### 17. Поиск подстроки вручную

Дана строка `text = "Hello world"` и подстрока `sub = "wor"`. Найдите индекс первого вхождения подстроки **вручную**: циклом проверяйте срез `text[i:i+len(sub)]` на каждом индексе. Не используйте `find`.

Тема: «Алгоритм поиска подстроки в строке».

### 18. else после цикла while

Дан список `a = [10, 20, 30, 40]`. Циклом `while` ищите в нём число 25. Если найдёте — выведите «найдено» и прервитесь. Блок `else` цикла должен вывести «не найдено», только если цикл завершился без `break`.

Тема: «else в цикле while».

### 19. Только чётные через continue

В цикле `for i in range(1, 11)` пропускайте нечётные числа оператором `continue` и выводите только чётные.

Тема: «Инструкция break, Оператор continue».

### 20. Таблица умножения (вложенные циклы)

Вложенными циклами `for` выведите таблицу умножения от 1 до 5: для каждой строки i выведите произведения i на 1..5 в одну строку. Используйте `end=" "` в `print` и пустой `print()` для перехода на новую строку.

Тема: «Вложенные циклы».

## Сложные задачи (10)

Алгоритмы сортировки, сдвига, поиска, матрицы, шифрование, тексты.

### 21. Сортировка пузырьком

Дан список `a = [5, 1, 4, 2, 8]`. Отсортируйте его **методом пузырька**: проходами сравнивайте соседние элементы и меняйте их местами, если левый больше правого. Выведите отсортированный список.

Тема: «Алгоритм сортировки списка (методом пузырька)».

### 22. Палиндром-строка

Дана строка `s = "level"`. Определите, является ли она палиндромом (читается одинаково слева направо и справа налево), сравнив её с перевёрнутой через срез `[:: -1]`, затем проверьте то же самое циклом по символам с двух концов.

Темы: «Строки», «Срезы», «Итерация по строке».

### 23. Удаление дубликатов без изменения порядка

Дан список `a = [1, 2, 3, 1, 2, 4, 3]`. Постройте новый список, в котором каждый элемент встречается один раз, сохранив порядок первого вхождения (элемент добавляется, если его ещё нет в новом списке — проверяйте через `in`).

Темы: «Списки», «Линейный поиск», «Добавление элемента».

### 24. Матрица: сумма элементов и главной диагонали

Дана квадратная матрица (список списков):

`m = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]`

Вложенными циклами: а) найдите сумму всех элементов; б) найдите сумму элементов главной диагонали (индексы `i`, `i`).

Тема: «Вложенные циклы».

### 25. Циклический сдвиг влево на k позиций

Дан список `a = [1, 2, 3, 4, 5]` и число `k = 2`. Выполните циклический сдвиг списка **влево** на `k` позиций: каждый элемент перемещается на k позиций назад, а первые k элементов переходят в конец. Реализуйте обобщённо (для любого k) с помощью цикла из алгоритма сдвига на 1, повторённого k раз.

Тема: «Алгоритм циклического сдвига в списке влево».

### 26. Все вхождения подстроки

Дана строка `text = "ababa"` и подстрока `sub = "aba"`. Посчитайте **все** вхождения подстроки, включая перекрывающиеся (они встречаются на индексах 0 и 2 — всего 2). Реализуйте циклом: от каждой позиции проверяйте срез `text[i:i+len(sub)]`. Не используйте `str.count` (он считает неперекрывающиеся).

Тема: «Алгоритм поиска подстроки в строке».

### 27. Простые числа до N

Для введённого `n` выведите все простые числа от 2 до `n` включительно. Число простое, если у него нет делителей кроме 1 и самого себя. Используйте вложенные циклы и `break`.

Темы: «Вложенные циклы», «Цикл for/while», «Инструкция break».

### 28. Шифр Цезаря

Реализуйте шифр Цезаря: для строки, состоящей из строчных букв латинского алфавита, сдвините каждую букву на `shift` позиций по алфавиту (алфавит замыкается в кольцо: после `"z"` снова `"a"`). Используйте `ord()` и `chr()` для перевода символа в код и обратно, как в лекции.

Пример: `"abc"` со сдвигом 2 даёт `"cde"`; `"xyz"` со сдвигом 2 даёт `"zab"`.

Темы: «Строки», «ord/chr» (из `Lesson_2.ipynb`), «Итерация по строке».

### 29. Разворот порядка слов

Дано предложение `s = "Быть или не быть"`. Разверните **порядок слов**: последнее слово становится первым и так далее (`"быть не или Быть"`). Разбейте строку через `split`, разверните список вручную (циклом или срезом) и склейте обратно f-строкой/оператором `+`.

Темы: «Методы строк (split)», «Обращение списка», «Итерация по списку».

### 30. Разворот букв в каждом слове

Дано предложение `s = "Hello world Python"`. Разверните буквы **внутри каждого слова**, сохранив порядок самих слов: `"olleH dlrow nohtyP"`. Разбейте строку через `split`, разверните каждое слово срезом `[:: -1]` и соберите результат обратно.

Темы: «Методы строк (split)», «Срезы», «Итерация по списку».
