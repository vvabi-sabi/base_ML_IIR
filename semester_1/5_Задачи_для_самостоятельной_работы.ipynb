# Задачи для самостоятельной работы

Лекция 5. **Ошибки и исключения, Декораторы, Генераторы**

Решите каждую задачу в ячейке под её условием. Задачи упорядочены по сложности: **лёгкие (10)**, **средние (10)**, **сложные (10)**. Все темы соответствуют слайдам лекции (см. `themes_5.txt`).

Правила оформления: следуйте PEP8 (4 пробела на отступ, осмысленные имена переменных).

## Лёгкие задачи (10)

Обработка исключений: базовый синтаксис `try: except:`, типы ошибок, блоки `else` и `finally`. Первые шаги с генераторами (`yield`, `next`) и простейшим декоратором.

### 1. Базовый синтаксис try / except

Попробуйте разделить `10 / 0` внутри блока `try`. Поймайте исключение `ZeroDivisionError` в блоке `except`, выведите сообщение и присвойте результату `None`. После блоков принтуйте результат.

Темы: «Базовый синтаксис (try: except:)», «Типы исключений (ZeroDivisionError)».

### 2. Типы исключений: IndexError и KeyError

Дан список `lst = [1, 2, 3]` и словарь `d = {'a': 1}`. В отдельных `try/except` попробуйте: обратиться к `lst[10]` (поймайте `IndexError`) и к `d['missing']` (поймайте `KeyError`). Выводите сообщение с текстом ошибки (`except ... as e`).

Темы: «Типы исключений (IndexError, KeyError)», «Базовый синтаксис (try: except:)».

### 3. Обработка нескольких типов исключений

Напишите функцию `convert(x)`, которая в `try` возвращает `int(x)`, а в `except` перехватывает **два типа сразу** — `(TypeError, ValueError)` — печатает сообщение и возвращает `None`. Проверьте её на `'123'`, `'abc'` (ValueError) и `None` (TypeError).

Темы: «Обработка нескольких типов исключений».

### 4. Исключение FileNotFoundError

Попробуйте открыть несуществующий файл `'no_such_file.txt'` в режиме чтения через `with open(...)`. Поймайте `FileNotFoundError`, выведите сообщение и присвойте `content = None`.

Темы: «Типы исключений (FileNotFoundError)», «Базовый синтаксис (try: except:)».

### 5. Исключение ValueError

Попробуйте преобразовать строку `s = 'abc'` в число через `int(s)`. Поймайте `ValueError`, выведите сообщение и задайте `num = 0`. Затем преобразуйте `'15'` без ошибки.

Темы: «Типы исключений (ValueError)», «Зачем обрабатывать исключения».

### 6. Блок else

Напишите функцию `divide(a, b)`: в `try` — вычисление `a / b`; в `except ZeroDivisionError` — вывод «деление на ноль» и возврат `None`; в блоке `else` — вывод «успешно» и возврат результата. Убедитесь, что блок `else` выполняется **только** при успехе.

Темы: «Блок else (после try: except:)».

### 7. Блок finally

Напишите функцию `process(a, b)`: в `try` — деление; в `except ZeroDivisionError` — `res = None`; в блоке `finally` — вывод «finally выполняется всегда». Проверьте и при успехе, и при ошибке: сообщение выводится в обоих случаях.

Темы: «Блок finally (после try: except:)».

### 8. Простейший генератор с yield

Напишите функцию-генератор `countdown(n)`, которая с помощью `while` и `yield` последовательно отдаёт числа от `n` до `1`. Выведите её элементы циклом `for`, затем превратите в список через `list()`.

Темы: «Определение итераторов и генераторов в Python (yield)».

### 9. Получение элементов через next()

Напишите генератор `evens(limit)`, который отдаёт чётные числа от 0 до `limit` (не включая). Создайте генератор `g = evens(6)` и получите первые три значения функцией `next(g)`.

Темы: «next(), send(), и т.д.», «Определение итераторов и генераторов (yield)».

### 10. Первый декоратор с @

Напишите декоратор `echo(func)`, внутренняя функция которого добавляет в список `log` метки `'start'` и `'end'` вокруг вызова исходной функции. Примените его к функции `hello()` через `@echo` и выведите `log`.

Темы: «Создание декоратора с помощью @», «Декораторы в Python».

## Средние задачи (10)

Оператор `raise`, собственные исключения, логирование, подходы LBYL и EAFP, генераторы с `next()` и остановкой `StopIteration`, декораторы с аргументами и возвратом значения, замыкания.

### 11. Оператор raise

Напишите функцию `positive_sqrt(x)`: если `x < 0`, она **преднамеренно** выбрасывает `ValueError` через `raise` (с сообщением «отрицательное число»), иначе возвращает `x ** 0.5`. Вызовите обе ветки, перехватив исключение.

Темы: «Использование оператора raise».

### 12. Собственное исключение

Создайте класс-исключение `NegativeNumberError`, унаследовав его от `Exception`. Напишите функцию `check_all(nums)`, которая выбрасывает это исключение при первом отрицательном числе (`raise NegativeNumberError(...)`). Перехватите исключение и проверьте, что оно поймано.

Темы: «Создание собственного исключения (пользовательский класс)».

### 13. Логирование исключений

Импортируйте модуль `logging` (настройте `logging.basicConfig(level=logging.INFO)`). Напишите функцию `safe_divide(a, b)`, которая возвращает результат деления, а при `ZeroDivisionError` вызывает `logging.exception('деление на ноль')` и возвращает `None`. Проверьте оба случая — при ошибке в журнал попадёт запись с трассировкой.

Темы: «Логирование исключений (import logging)», «Обработка всех типов исключений».

### 14. Подходы LBYL и EAFP

Реализуйте деление с проверкой двумя подходами: (1) **LBYL** — сначала проверить `b != 0`, потом делить; (2) **EAFP** — сразу пробовать делить в `try`, а `ZeroDivisionError` обрабатывать в `except`. Обе возвращают `None` при делении на ноль. Сравните результаты.

Темы: «Подход LBYL, Подход EAFP (сравнение)».

### 15. Генератор чисел Фибоначчи

Напишите **бесконечный** генератор `fib()`, который через `yield` отдаёт числа Фибоначчи (0, 1, 1, 2, ...) без ограничения. С помощью `next()` получите первые 10 чисел в цикле и выведите их.

Темы: «Определение итераторов и генераторов (yield)», «Полезные примеры».

### 16. next() и исчерпание генератора (StopIteration)

Напишите генератор `squares(limit)`, отдающий квадраты чисел от 1 до `limit`. Получите значения функцией `next()` в цикле `while True`, а когда генератор исчерпается — перехватите `StopIteration`, выведите «генератор исчерпан» и выйдите циклом `break`.

Темы: «next(), send(), и т.д.».

### 17. Декоратор для функций с аргументами

Напишите декоратор `log_args`, внутренняя функция которого принимает `*args` и `**kwargs`, печатает их и вызывает исходную функцию с теми же аргументами (`func(*args, **kwargs)`). Примените к функции `add(a, b)` и вызовите её позиционно и по именам.

Темы: «Декорирование функций, принимающих аргументы (*args, **kwargs)».

### 18. Возврат значения из декоратора

Напишите декоратор `double_result`, внутренняя функция которого **возвращает** результат исходной функции, умноженный на 2 (не забудьте `return`). Примените к функции `get_num()` и убедитесь, что декорированная функция возвращает число (иначе был бы `None`).

Темы: «Возвращение значения из декорированных функций».

### 19. Декоратор, принимающий аргументы

Напишите фабрику декораторов `repeat(times)`: она возвращает декоратор, который вызывает исходную функцию `times` раз подряд. Примените `@repeat(3)` к функции `ping()`, которая добавляет элемент в список `counts`, и проверьте, что после вызова в списке 3 элемента.

Темы: «Декораторы, принимающие аргументы».

### 20. Внутренние функции и замыкание

Напишите функцию `make_multiplier(n)`, которая **возвращает** внутреннюю функцию `multiplier(x)`, умножающую `x` на `n` (внутренняя функция «замыкает» переменную `n`). Создайте `double` и `triple`, выведите примеры.

Темы: «Внутренние функции», «Возврат функции из функции», «Замыкания».

## Сложные задачи (10)

`functools.wraps`, декоратор тайминга, несколько декораторов, корутины (`value = yield`, `send()`), машина конечных состояний, рекурсивные генераторы (`yield from`), отладочный декоратор (число e), сравнение времени генератора и списка, генераторы для файлов, декораторы классов.

### 21. functools.wraps: сохранение имени и docstring

Создайте декоратор `my_decorator`, внутренняя функция которого помечена `@wraps(func)` (импортируйте `functools.wraps`). Примените его к функции `greet(name)` с docstring, проверьте, что сохранились `__name__` и `__doc__`. Для сравнения сделайте декоратор **без** `@wraps` и покажите, что имя теряется (становится `wrapper`).

Темы: «Интроспекция имени декорируемой функции (functools.wraps)».

### 22. Декоратор для тайминга кода

Напишите декоратор `timer`, который замеряет время выполнения функции через `time.time()`, печатает имя функции и затраченное время (`.4f` сек) и **возвращает** результат. Примените к функции `slow_sum(n)`, суммирующей числа от 0 до `n-1` циклом, и вызовите для `n = 1000000`.

Темы: «Декоратор для тайминга кода», «Сравнение времени работы».

### 23. Применение нескольких декораторов

Напишите два декоратора: `shout` (приводит строковый результат к верхнему регистру) и `add_marks` (печатает `>>` перед вызовом и `<<` после). Примените их к одной функции друг над другом (`@add_marks` над `@shout`) и объясните, что применяются «снизу вверх»: сначала `shout`, потом `add_marks`.

Темы: «Применение нескольких декораторов».

### 24. Корутина: value = yield и send()

Напишите функцию-генератор `averager()`, работающую как корутина: внутри используется конструкция `value = yield avg`, где `avg` — текущее среднее. «Раскрутите» корутину первым `next()`, затем отправляйте числа через `send(value)` и получайте текущее среднее (10 → 10.0, затем 20 → 15.0, затем 30 → 20.0).

Темы: «Итераторы как корутины (value = yield)», «next(), send(), и т.д.».

### 25. Машина конечных состояний на генераторе

Реализуйте с помощью `yield` машину конечных состояний «светофор»: состояния `red → green → yellow → red → ...`. Напишите генератор `traffic_light()`, который при каждом `next()` отдаёт текущее состояние и переключает его. Получите 6 состояний и проверьте последовательность.

Темы: «Машина конечных состояний (определение)», «Машина конечных состояний (с использованием yield)».

### 26. Рекурсивный генератор с yield from

Напишите генератор `flatten(nested)`, который «разворачивает» вложенные списки любой глубины. Для вложенного списка вызывайте себя через `yield from flatten(item)`, для обычного элемента — `yield item`. Проверьте на структуре `[1, [2, [3, 4], 5], 6]`.

Темы: «Рекурсивный вызов генератора (yield from)», «Примеры».

### 27. Отладочный декоратор (число e)

Напишите отладочный декоратор `debug`, который печатает имя функции, её аргументы и результат (`print(f'{func.__name__}({args}) = {result}')`) и **возвращает** результат. Примените его к функции `e_series(n)`, вычисляющей сумму первых `n` членов ряда для числа e (`1/0! + 1/1! + 1/2! + ...`, где `k!` — факториал). Убедитесь, что при большом `n` сумма близка к `math.e`.

Темы: «Отладочный декоратор (вычисление числа e)», «Возвращение значения из декорированных функций».

### 28. Сравнение времени: список vs генератор

Посчитайте сумму чисел от 0 до 999999 двумя способами: (1) создав настоящий список `list(range(n))` и просуммировав его; (2) через генератор, который отдаёт числа по одному. Замерьте время каждого способа функцией `time.time()` и убедитесь, что суммы совпадают. Обсудите экономию памяти генератора.

Темы: «Сравнение времени работы», «Определение итераторов и генераторов (yield)».

### 29. Генератор для чтения файла построчно

Напишите генератор `read_lines(path)`, который открывает файл через `with` и отдаёт строки по одной через `yield` (позволяет обрабатывать большие файлы без загрузки в память). Создайте файл `sample.txt` с тремя словами, «прочитайте» его генератором и выведите строки с номерами.

Темы: «Полезные примеры (генераторы)», «Итерируемый объект».
