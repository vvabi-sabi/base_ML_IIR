## Лёгкие задачи (10)

Кортежи, словари и их методы, множества, функции (`def`, `return`, `pass`), аргументы функций.

### 1. Создание и индекс кортежа

Создайте кортеж `t = (10, 20, 30, "abc")`. Выведите: первый элемент, последний элемент (индекс `-1`), длину через `len()`. Затем попробуйте изменить элемент (`t[0] = 100`) и объясните/перехватите ошибку.

Темы: «Кортежи, отличие от списков», «Создание кортежа», «Неизменяемость кортежей».

### 2. Сложение и дублирование кортежей

Даны кортежи `a = (1, 2)` и `b = (3, 4)`. Склейте их оператором `+` и продублируйте результат умножением `*`. Выведите оба результата.

Темы: «Кортежи», «Создание кортежа (операции + и *)».

### 3. Словарь: создание и доступ по ключу

Создайте словарь `student = {'name': 'Аня', 'age': 20, 'city': 'Новосибирск'}` двумя способами: через `{}` и через `dict(...)`. Выведите значение по ключу `'name'` через квадратные скобки и через метод `.get()`. Выведите число пар через `len()`.

Темы: «Словари, создание», «Методы словарей (.get)».

### 4. Методы словаря: keys, values, items

Дан словарь `d = {'a': 1, 'b': 2, 'c': 3}`. Выведите все ключи (`.keys()`), все значения (`.values()`) и все пары (`.items()`), преобразовав их в список через `list()`. Найдите сумму всех значений функцией `sum(d.values())`.

Темы: «Методы словарей (keys, values, items)».

### 5. Итерация по словарю

Дан словарь `ages = {'Иван': 25, 'Мария': 30, 'Пётр': 28}`. Выведите: (а) все ключи циклом `for key in ages`; (б) все пары «ключ: значение» циклом по `ages.items()` через f-строку.

Темы: «Итерация по словарю (по ключам, значениям, парам)».

### 6. Множества: создание и уникальность

Дан список `nums = [1, 2, 2, 3, 3, 3]`. Создайте из него множество через `set()`. Выведите результат и проверьте, что дубликаты исчезли. Проверьте принадлежность элемента 2 оператором `in`, затем добавьте элемент 4 методом `add()`.

Темы: «Множества в Python», «Методы множеств (add)», «Оператор in».

### 7. Операции над множествами

Даны множества `a = {1, 2, 3, 4}` и `b = {3, 4, 5, 6}`. Вычислите и выведите: объединение (`|`), пересечение (`&`), разность (`-`), симметрическую разность (`^`).

Темы: «Операции над множествами (|, &, -, ^)».

### 8. Функция с return и функция-заглушка

Напишите функцию `squared(n)`, которая возвращает квадрат числа через `return`. Вызовите её для 5 и 7. Затем напишите функцию `stub()` с телом `pass` и вызовите её.

Темы: «Функции. Аргументы функции, возвращаемое значение», «Функции с pass».

### 9. Аргументы по умолчанию

Напишите функцию `greet(name, greeting='Привет')`, которая выводит f-строку `'{greeting}, {name}!'`. Вызовите её один раз без второго аргумента и второй раз с другим приветствием.

Темы: «Задание аргументов по умолчанию», «f-строки».

### 10. Именованные аргументы в произвольном порядке

Напишите функцию `show(name, age, city)`, которая выводит `f'{name}, {age} лет, г. {city}'`. Вызовите её, передавая аргументы по именам в **произвольном порядке** (например, `show(city='Москва', age=20, name='Аня')`).

Темы: «Вызов функции со значениями в произвольном порядке».

## Средние задачи (10)

### 11. Частотный словарь

Дан список слов `words = ['яблоко', 'груша', 'яблоко', 'слива', 'груша', 'яблоко']`. Постройте частотный словарь: ключ — слово, значение — сколько раз оно встречается. Используйте цикл `for` и метод `.get(key, 0)` со значением по умолчанию.

Темы: «Словари», «Метод get с запасным вариантом», «Итерация по списку».

### 12. Значение по ключу с запасным вариантом

Дан словарь столиц `capitals = {'Россия': 'Москва', 'Франция': 'Париж', 'Германия': 'Берлин'}`. Для списка стран `countries = ['Россия', 'Италия', 'Франция']` выведите столицу каждой страны; если страны нет, выведите `'Не знаю'`. Используйте `.get(key, 'Не знаю')`.

Темы: «Методы словарей (.get)», «Словари, создание».

### 13. *args: сумма и среднее

Напишите функцию `stats(*args)`, которая принимает произвольное количество чисел и возвращает кортеж `(сумма, среднее)`. Внутри выведите сам кортеж `args`. Вызовите с тремя числами, с пятью числами и без аргументов (для пустого ввода верните среднее 0).

Темы: «Произвольное число аргументов (*args)».

### 14. **kwargs: профиль пользователя

Напишите функцию `profile(**kwargs)`, которая выводит все пары «ключ: значение», переданные как именованные аргументы. Вызовите: `profile(name='Аня', age=20, city='Новосибирск')`.

Темы: «Произвольное число именованных аргументов (**kwargs)».

### 15. lambda-функция в sorted

Дан список кортежей `students = [('Аня', 20), ('Борис', 19), ('Вера', 22)]`. Отсортируйте его по возрасту (второму элементу) через `sorted(..., key=lambda x: x[1])`, затем по имени (`x[0]`). Выведите оба результата.

Темы: «Анонимные функции (lambda)», «Сортировка sorted с key».

### 16. Рекурсия: факториал

Напишите рекурсивную функцию `factorial(n)` с базовым случаем (`n == 0`) и рекурсивным шагом (`n * factorial(n - 1)`). Добавьте короткий docstring. Вычислите `5!` и `10!`.

Темы: «Рекурсия (базовый случай, рекурсивный шаг)», «Docstring».

### 17. Рекурсия: сумма от 1 до n

Напишите рекурсивную функцию `sum_to(n)`, которая возвращает сумму чисел от 1 до `n`. Проверьте на `n = 100` (ожидаемый результат 5050) и сравните с формулой `n * (n + 1) // 2`.

Темы: «Рекурсия», «Возвращаемое значение функции».

### 18. Кэширование чисел Фибоначчи

Реализуйте функцию `fib(n)` с кэшем — словарём `M = {0: 0, 1: 1}`, в который сохраняются уже вычисленные значения (как в лекции). Вычислите список первых 16 чисел Фибоначчи и выведите заполненный кэш `M`.

Темы: «Кэширование функций (на примере „Числа Фибоначчи“)», «Словари».

### 19. Файлы: запись и чтение

Создайте в **текущей папке** файл `test.txt` в режиме записи `'w'` и запишите в него три строки через цикл. Затем откройте его в режиме чтения `'r'` и выведите всё содержимое методом `read()` (не забудьте про `encoding='utf-8'` и конструкцию `with`).

Темы: «Работа с файлами. Режим открытия (r, w)», «Методы файла (f.write, f.read)», «Файловый ввод/вывод».

### 20. Файлы: режим append и построчное чтение

Откройте файл `log.txt` в режиме добавления `'a'` и допишите строку `'New entry'`. Затем откройте его в режиме чтения и выведите содержимое построчно — циклом `for line in f`. Запустите ячейку дважды и убедитесь, что строки дописываются, а не перезаписываются.
