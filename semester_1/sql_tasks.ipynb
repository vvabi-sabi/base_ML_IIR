## Задачи по SQLite для Python (от простой к сложной)

### Уровень 1: Основы (Создание, Вставка, Чтение)

```python
Задача 1: "Знакомство"
Создайте новую базу данных library.db. Внутри базы создайте таблицу books со следующими полями:
*   id (INTEGER, PRIMARY KEY)
*   title (TEXT, NOT NULL)
*   author (TEXT, NOT NULL)
*   year (INTEGER)

Цель: Научиться подключаться к БД, выполнять простой SQL-запрос на создание таблицы.
```

```python
Задача 2: "Пополнение полки"
Напишите скрипт, который добавляет 5 различных книг в таблицу books. Сделайте это двумя способами:
1.  Выполните пять отдельных запросов INSERT.
2.  Выполните один запрос INSERT, используя механизм подстановки параметров (?) и передав список кортежей.

Цель: Освоить вставку данных, понять важность и синтаксис подстановки параметров для безопасности.
```

```python
Задача 3: "Каталог"
Напишите скрипт, который получает и выводит на экран все книги из таблицы books тремя способами:
1.  Получить все записи с помощью fetchall().
2.  Получить записи по одной с помощью fetchone() в цикле.
3.  Получить первые 3 записи с помощью fetchmany(3).

Цель: Изучить различные методы получения данных из курсора.
```
