# Задачи для самостоятельной работы

Лекция 6. **Работа с текстовыми данными**

Решите каждую задачу в ячейке под её условием. Задачи упорядочены по сложности: **лёгкие (10)**, **средние (10)**, **сложные (10)**. Все темы соответствуют слайдам лекции (см. `themes_6.txt`).

Правила оформления: следуйте PEP8 (4 пробела на отступ, осмысленные имена переменных).

## Лёгкие задачи (10)

Строки и их методы, наивный поиск подстроки, основы регулярных выражений (`re`: search/fullmatch, односимвольные шаблоны, квантификаторы), кодирование строк UTF-8, простое RLE-сжатие, one-hot кодирование слова и базовая токенизация.

### 1. Стандартные методы строк

Возьмите строку `'  Hello, World!  '` (с пробелами по краям). Продемонстрируйте методы: `len`, `lower`, `upper`, `strip`, `replace`, `count`, `startswith`, `endswith`, а также `split` и `join`. Выведите результаты и проверьте ключевые через `assert`.

Темы: «Стандартные методы и функции для строк (python)».

### 2. Строковые функции и проверки

Продемонстрируйте проверки `isdigit`, `isalpha`, `isalnum`, преобразования `title`, `capitalize`, методы поиска `find`, `index`, `rfind`, а также **срезы** строк (обратный порядок, подстрока). Выведите результаты и проверьте через `assert`.

Темы: «Стандартные методы и функции для строк (python)».

### 3. Наивный поиск подстроки

Напишите функцию `naive_search(text, pattern)`, которая возвращает список всех позиций вхождения `pattern` в `text` перебором всех возможных начальных позиций. Проверьте на примерах из лекции («abracadabra» / «abra»), включая перекрывающиеся вхождения. Сравните результат со встроенным `str.find` в цикле.

Темы: «Поиск подстроки в строке (наивный алгоритм)».

### 4. One-hot кодирование слова

Дан словарь `vocab` из восьми слов (как в лекции: time, fruit, flies, like, a, an, arrow, banana). Напишите функцию `one_hot(word, vocab)`, возвращающую унитарный вектор: единица на позиции слова, нули в остальных. Выведите векторы для нескольких слов и проверьте через `assert`.

Темы: «One-hot кодирование».

### 5. Токенизация текста

Напишите функцию `tokenize(text)`, которая разбивает текст на слова: приведите к нижнему регистру и найдите все последовательности букв (и апострофов) с помощью `re.findall(r"[a-zA-Z']+", text)`. Для сравнения реализуйте вариант без регулярных выражений (замените знаки пунктуации на пробелы и вызовите `split`). Проверьте через `assert`.

Темы: «Токенизация».

### 6. Знакомство с модулем re: search, fullmatch, findall

Напишите код, который в строке `'Мой номер телефона: +7 (913) 123-45-67, позвоните мне.'` выполняет: `re.search(r'\d+')` (первое число и его позиция), `re.fullmatch` для строк `'123'` и `'12a3'`, и `re.findall(r'\d+')` (все числа). Выведите результаты и проверьте через `assert`.

Темы: «Регулярные выражения в Python (определение)», «Основы синтаксиса», «Регулярки в python (модуль re)».

### 7. Односимвольные шаблоны

Продемонстрируйте односимвольные шаблоны из лекции: `\d` (цифра), `\D` (не цифра), `\w` (буква), `\s` (пробельный символ), `.` (любой символ кроме перевода строки). Используйте `re.findall` на подходящих строках и проверьте, что `r'м.л.ко'` находит «молоко» и «малако».

Темы: «Односимвольные шаблоны».

### 8. Квантификаторы

В строке `'Цены: 5, 99, 1000 рублей. Код СУ35, ещё СУ-111.'` выделите числа с помощью `\d` и `\d+`, объясните разницу. Продемонстрируйте `\d{2}`, `\d{1,3}` и `colou?r` (вариант `?` — 0 или 1 символ). Проверьте через `assert`.

Темы: «Квантификаторы (количество повторений)».

### 9. Простое RLE-кодирование

Метод сжатия RLE (Run-Length Encoding) заменяет серию одинаковых символов на символ и количество повторов. Напишите функцию `rle_encode(s)`, которая превращает `'AAABBC'` в `'A3B2C1'`. Проверьте на нескольких строках через `assert`.

Темы: «Сжатие строки», «Кодирование длины прогона (RLE)».

### 10. Кодирование строк: UTF-8

Закодируйте строку `'Привет, мир!'` в байты (`str.encode('utf-8')`), выведите байты и hex-представление, затем восстановите исходную строку (`bytes.decode`). Сравните число символов и число байтов (кириллица занимает 2 байта в UTF-8). Проверьте, что буква «А» в UTF-8 — это `d090` в hex.

Темы: «Кодирование строк (UTF-8)».

## Средние задачи (10)

Расстояние Левенштейна, Z-функция и поиск подстроки с её помощью, полное RLE (кодирование + декодирование), скобочные группы для валидации MAC-адреса, поиск Боуера—Мура, эвристический стемминг, TF-векторизация предложения, бинарное (свёрнутое) кодирование, токенизация с очисткой текста.

### 11. Расстояние Левенштейна

Напишите функцию `levenshtein(a, b)` — редакционное расстояние между строками через динамическое программирование (таблица `dp`, операции вставки/удаления/замены). Проверьте на примерах: `levenshtein('kitten', 'sitting') == 3`, равенство даёт 0 и т.д.

Темы: «Расстояние Левенштейна (редакционное расстояние)».

### 12. Z-функция

Напишите функцию `z_function(s)`, возвращающую Z-массив: `z[i]` — длина наибольшего общего префикса строки `s` и её суффикса, начинающегося с позиции `i` (`z[0] = 0`). Проверьте на примерах из лекции: `'aaaaa'`, `'abacaba'`, `'abracadabra'`.

Темы: «Z-строки (Z-функция)».

### 13. Поиск подстроки через Z-функцию

Используя Z-функцию, реализуйте поиск всех вхождений образца `p` в текст `t` способом из лекции: строка `s = p + '$' + t` (разделитель, которого нет в тексте), затем по Z-массиву ищем позиции, где `z[i] >= len(p)`. Сравните с наивным поиском.

Темы: «Z-строки», «Применение Z-функции (поиск подстроки)».

### 14. RLE: кодирование и декодирование

Дополните задачу 9: напишите `rle_decode(s)`, восстанавливающую строку из RLE-представления (`'A4B3C2D1A2'` → `'AAAABBBCCDAA'`). Убедитесь, что `rle_decode(rle_encode(s)) == s` для нескольких строк.

Темы: «Сжатие строки», «Кодирование длины прогона (RLE)».

### 15. Скобочные группы: MAC-адрес

Запишите регулярное выражение для проверки MAC-адреса вида `01:23:45:67:89:ab` (шесть групп из двух шестнадцатеричных цифр, разделитель `:` или `-`) с помощью несохраняющей группы `(?:...)` и квантификатора `{5}`: `r'(?:[0-9a-fA-F]{2}[:-]){5}[0-9a-fA-F]{2}'`. Напишите функцию `is_valid_mac(a)` с `re.fullmatch` и проверьте набор адресов.

Темы: «Скобочные группы и пересечения», «Регулярки в python (модуль re)».

### 16. Эвристический стемминг

Напишите функцию `simple_stem(word)`, реализующую простой стемминг: последовательно отбрасывайте типовые окончания (например, `'ing', 'ed', 'es', 's', 'ly', 'ion', 'ment', 'ness'`), если после отбрасывания остаётся не меньше трёх символов. Проверьте на словах `'running'`, `'cats'`, `'quickly'`, `'played'`.

Темы: «Стемминг».

### 17. TF-векторизация предложения

Частота термина (TF) для предложения — это вектор по словарю, в котором на позиции слова стоит количество его вхождений. Дан словарь из лекции. Напишите `tf_vector(text, vocab)` через `Counter` и вычислите векторы для предложений со словами из лекции. Проверьте через `assert`.

Темы: «Частотные модели языка», «TF-представление», «Векторизация предложений».

### 18. Бинарное (свёрнутое) кодирование

Свёрнутое бинарное кодирование фразы — вектор длины словаря, где 1 означает наличие слова во фразе (повторы учитываются один раз). Напишите `binary_vector(text, vocab)` через `set` и проверьте пример из лекции: `'like a banana'` → `[0, 0, 0, 1, 1, 0, 0, 1]`.

Темы: «One-hot кодирование», «Свёрнутое (бинарное) кодирование».

### 19. Поиск Боуера—Мура: таблица смещений

Реализуйте упрощённый поиск Боуера—Мура: образец сравнивается с текстом **справа налево**, а при несовпадении используется таблица смещений «по плохому символу» (для каждого символа образца — на сколько можно сдвинуться). Напишите `build_shift_table(pattern)` и `boyer_moore(text, pattern)`. Проверьте на примерах.

Темы: «Алгоритм Р. Боуера и Д. Мура (БМ-поиск)».

### 20. Токенизация с очисткой и подсчёт частот

Возьмите текст `'The quick brown fox jumps over the lazy dog. A fox and a dog are friends.'`. Разбейте его на слова (`re.findall`), уберите **стоп-слова** (`a, an, the, and, of, in, on, to, is, are`) и посчитайте частоты оставшихся слов через `Counter`. Проверьте через `assert`.

Темы: «Токенизация», «Частотные модели языка».

## Сложные задачи (10)

Левенштейн с восстановлением пути, сжатие строки через минимальный период (Z-функция), лематизация, TF-IDF, N-граммы, валидация регистрационных знаков, контекстные (семантические) векторы в духе Word2vec, классификация спама на основе «мешка слов», улучшенный БМ-поиск и полный конвейер обработки текста.

### 21. Расстояние Левенштейна с восстановлением пути

Дополните задачу 11: функция `levenshtein_with_ops(a, b)` должна возвращать не только расстояние, но и список операций (удаление/вставка/замена), восстанавливаемый по таблице `dp` с конца. Для `'kitten' -> 'sitting'` расстояние 3 и 3 операции. Выведите пошагово операции.

Темы: «Расстояние Левенштейна (редакционное расстояние)».

### 22. Сжатие строки через минимальный период (Z-функция)

Дана строка `s`. Найдите длину наименьшей строки `t`, такой что `s` — это несколько копий `t` подряд (конкатенация). Используйте Z-функцию: строку можно сжать до периода `p` (p делит длину и `z[p] >= n - p`). Проверьте на `'abcabcabc'`, `'aaaa'`, `'abababab'` и несжимаемой строке.

Темы: «Z-строки», «Сжатие строки (минимальный период)».

### 23. Простая лематизация (эвристика)

Напишите функцию `simple_lemmatize(word)`, которая приводит слово к начальной форме: сначала проверяет словарь исключений (`'люди' -> 'человек'`, `'лучше' -> 'хорошо'`), затем отбрасывает типовые окончания. Проверьте на формах слова «стол» (`стола`, `столу`, `столом`, `столами`) и исключениях. Обратите внимание: это учебная эвристика — на практике используют pymorphy2/nltk.

Темы: «Лематизация».

### 24. TF-IDF для нескольких документов

Реализуйте меру TF-IDF. Для трёх текстов: 1) посчитайте TF (частота слова / длина документа), 2) IDF (log(числа документов / числа документов со словом)), 3) постройте TF-IDF вектор документа. Проверьте, что у слова `'the'` (встречается в 2 из 3 документов) IDF ≈ 0.405, а у `'cat'` (в 1 из 3) ≈ 1.099.

Темы: «Частотные модели языка», «TF-IDF», «Векторизация предложений».

### 25. N-граммы

Напишите функции `char_ngrams(s, n)` (последовательности из n символов) и `word_ngrams(s, n)` (из n слов подряд). Продемонстрируйте биграммы и триграммы на предложении, постройте частотный словарь биграмм через `Counter` и выведите самые частые. Проверьте через `assert`.

Темы: «Частотные модели языка», «N-граммы».

### 26. Проверка регистрационных знаков (задача из лекции)

В лекции приведён шаблон для валидации номерных знаков: `^[A-Z]{1,3}-\d{1,4}$` (1–3 прописные буквы, дефис, затем 1–4 цифры). Напишите функцию `valid_plate(s)` с `re.fullmatch` и проверьте набор строк: корректные (`A-123`, `ABC-1234`, `AB-12`) и некорректные (`A-12345`, `12-AB`, `abc-123`, `A--12`).

Темы: «Задача: Регистрационные знаки», «Регулярки в python (модуль re)», «Скобочные группы».

### 27. Контекстные векторы (идея Word2vec)

Word2vec опирается на идею: слова в похожих контекстах имеют похожие представления. Постройте «контекстные» векторы для корпуса из трёх предложений: для каждого слова посчитайте частоты слов в окне ±2 символа вокруг него. Напишите `context_vector(corpus)` и функцию сходства двух слов, проверьте, что «яблоко» ближе к «апельсин», чем слово вне корпуса.

Темы: «Word2vec», «Семантика, чем определяется смысл слов», «Частотные модели языка».

### 28. «Мешок слов» и классификация спама

Постройте простой классификатор на основе Bag of Words. По обучающим документам (`spam_docs`, `ham_docs`) подсчитайте, сколько раз каждое слово встречалось в спаме и в обычных письмах. Для нового текста классифицируйте: «spam», если сумма (частота в спаме − частота в «хэме») положительна. Проверьте на примерах.

Темы: «Модель „Мешок слов“», «Частотные модели языка».

### 29. Улучшенный БМ-поиск: проверка согласованности

Доработайте БМ-поиск из задачи 19. Добавьте таблицу с **последним** вхождением символа и функцию, которая находит все вхождения. Главное требование — результаты должны **совпадать с наивным поиском** на наборе примеров (`'test test test'` / `'test'` → [0, 5, 10]). Выведите сдвиги по таблице и позиции.

Темы: «Алгоритм Р. Боуера и Д. Мура (БМ-поиск)».

### 30. Полный конвейер: токенизация → стемминг → BoW → косинусная близость

Соберите полный конвейер обработки текста: 1) токенизация (`re.findall`), 2) стемминг, 3) удаление стоп-слов, 4) мешок слов (`bow`), 5) косинусная близость векторов. Проверьте, что перестановка слов в одном тексте даёт высокое сходство (BoW не учитывает порядок слов), а совсем другой текст — низкое.

Темы: «Токенизация», «Стемминг», «Мешок слов», «Векторизация предложений».
