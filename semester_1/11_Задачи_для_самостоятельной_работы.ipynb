# Лекции 10–11 — Классы и ООП: задачи для самостоятельной работы

**Темы:** класс, `__init__` и `self`, атрибуты класса и экземпляра, `@property` и сеттеры, `getattr`/`setattr`, магические методы (`__repr__`, `__eq__`, `__len__`, `__getitem__`, `__iter__`, `__add__`, `__call__`, `__contains__`), наследование и `super()`, полиморфизм, MRO, `__slots__`, `__init_subclass__`, дескрипторы, мини-ORM, класс-матрица, стек и очередь, сериализация в JSON, Observer, registry/фабрика.

- **Задачи 1–10** — простые
- **Задачи 11–20** — средние
- **Задачи 21–30** — сложные

Все решения должны работать **без интернета** и без сторонних библиотек (только стандартная библиотека).


## Простые задачи


### 1. Первый класс
Создайте класс `Dog`: метод `__init__(self, name, age)` сохраняет `self.name` и `self.age`, а метод `say()` возвращает строку вида `Rex is 3 years old`. Создайте двух собак и вызовите у обеих `say()`.

**Цель:** `Dog("Rex", 3).say() == "Rex is 3 years old"`.

**Подсказка:** первый аргумент любого метода экземпляра — `self`; при вызове `Dog("Rex", 3)` Python сам передаёт новый объект в `__init__`.


### 2. Атрибут класса vs атрибут экземпляра
В классе `Robo` создайте атрибут класса `count = 0` и увеличивайте его в `__init__` при каждом создании робота. Создайте три экземпляра и проверьте `Robo.count`. Затем присвойте `r1.count = 100` и объясните, почему у класса значение не изменилось.

**Цель:** после трёх экземпляров `Robo.count == 3`; после `r1.count = 100` имеем `r1.count == 100`, но `Robo.count == 3`.

**Подсказка:** Python ищет атрибут сначала в экземпляре, потом в классе. Запись `self.count += 1` создаёт атрибут экземпляра, а не меняет класс.


### 3. Свойство @property
Класс `Temperature` хранит `self.celsius`. Сделайте свойство `fahrenheit` (геттер через `@property`), вычисляемое по формуле `c * 9 / 5 + 32`. Меняйте `celsius` — свойство должно отдавать актуальное значение.

**Цель:** `Temperature(100).fahrenheit == 212`, а после `t.celsius = 0` — `t.fahrenheit == 32`.

**Подсказка:** метод, decorated `@property`, вызывается как обычный атрибут, без скобок.


### 4. Свойство с сеттером
В классе `Person` сделайте `age` свойством: геттер возвращает `self._age`, сеттер бросает `ValueError`, если возраст отрицательный или больше 150. В `__init__` присваивайте возраст через свойство (`self.age = age`).

**Цель:** после `p.age = -1` возвышается `ValueError`, а старое значение сохраняется.

**Подсказка:** второй метод с тем же именем и декоратором `@age.setter`; внутри него уже доступен `value`.


### 5. getattr и setattr
Создайте пустой класс `Bag` (`pass`). Из словаря `{"name": "Ann", "age": 30}` назначьте все поля объекту через `setattr(obj, key, value)`. Затем прочитайте поля через `getattr` и обратитесь к несуществующему полю с значением по умолчанию.

**Цель:** `getattr(obj, "name") == "Ann"`, `getattr(obj, "weight", 0) == 0`, `hasattr(obj, "age") is True`.

**Подсказка:** `setattr`/`getattr` работают с именем-строкой, поэтому имена можно брать прямо из ключей словаря.


### 6. __str__ и __repr__
Класс `Point(x, y)`. Реализуйте `__str__` для человека — `Point (3, 4)`, и `__repr__` для программиста — `Point(x=3, y=4)`, так чтобы `eval(repr(p))` воссоздавал точку.

**Цель:** `repr(Point(3, 4)) == "Point(x=3, y=4)"` и `eval(repr(p))` даёт равные координаты.

**Подсказка:** `repr` — «исходник» объекта, `str` — его человекочитаемое описание; f-строки упрощают оба метода.


### 7. Сравнение объектов: __eq__
Класс `Book(title, author)`. По умолчанию два разных объекта не равны между собой. Реализуйте `__eq__`: книги равны, когда совпадают `title` и `author`.

**Цель:** `Book("Dune", "Herbert") == Book("Dune", "Herbert")` и `Book("Dune", "Herbert") != Book("Dune", "Strugatsky")`.

**Подсказка:** без `__eq__` Python сравнивает `is` (идентичность). Возвращайте `NotImplemented` для объектов чужого типа.


### 8. Магический метод __len__
Класс `Playlist`: список треков внутри, метод `add(track)` и `__len__`, возвращающий число треков.

**Цель:** после двух `add` вызов `len(pl)` возвращает `2`.

**Подсказка:** встроенный `len(x)` — это синтаксический сахар над `x.__len__()`.


### 9. Класс-стек
Реализуйте класс `Stack`: `push`, `pop`, `peek` (верхний без снятия), `is_empty` и `__len__`. `pop` с пустого стека должен бросать `IndexError` с внятным сообщением.

**Цель:** после трёх `push` и одного `pop` остаются два элемента в правильном порядке (LIFO); `Stack().pop()` бросает `IndexError`.

**Подсказка:** список-обёртка: `append` для push и `pop()` для pop — оба работают за O(1) в конец списка.


### 10. Приватные атрибуты
Класс `Account(owner, balance=0)`: поле `__balance` сделайте «приватным», добавьте методы `deposit(amount)` и `withdraw(amount)` с проверками (сумма положительна, снятие не больше баланса) и свойство `balance` только для чтения.

**Цель:** `acc.balance == 150` после депозита; снятие большей суммы бросает `ValueError`; `acc.__balance` бросает `AttributeError`, но `acc._Account__balance` доступен.

**Подсказка:** два подчёркивания в имени включают name mangling: поле сохраняется как `_Account__balance`.


## Средние задачи


### 11. __getitem__ и __contains__
Класс `Bookshelf` хранит список книг. Реализуйте `__getitem__` (по индексу и по срезу), `__contains__` для оператора `in` и `__len__`.

**Цель:** `shelf[0] == "Dune"`, `shelf[1:3]` возвращает список, `"Dune" in shelf` равно `True`.

**Подсказка:** в `__getitem__` ключ приходит готовым — список сам понимает и `int`, и `slice`.


### 12. Итератор: __iter__
Класс `Countdown(n)` — обратный счётчик: `list(Countdown(3))` должно давать `[3, 2, 1]`. Реализуйте `__iter__` через генератор или через `__next__` с `StopIteration`.

**Цель:** `list(Countdown(5)) == [5, 4, 3, 2, 1]`, объект можно передать в `sum()`.

**Подсказка:** самый простой `__iter__` — тот, который содержит `yield`.


### 13. Арифметика: __add__ и __mul__
Класс `Vector2D(x, y)`: сложение векторов через `+`, умножение на число через `*` (оба оператора возвращают **новый** вектор, а старые не меняются). Добавьте `__eq__`, `__repr__` и свойство `coords`.

**Цель:** `(Vector2D(1, 2) + Vector2D(3, 4)).coords == (4, 6)` и `(Vector2D(1, 2) * 2).coords == (2, 4)`.

**Подсказка:** операторные методы не должны мутировать `self` — конструируйте и возвращайте новый объект.


### 14. Вызываемый объект: __call__
Класс `CallCounter`: каждый вызов экземпляра как функции увеличивает счётчик и возвращает текущее число вызовов. Вторым классом сделайте `Multiplier(k)` — экземпляр умножает аргумент на `k`.

**Цель:** после трёх вызовов счётчик вернул `3`; `Multiplier(3)(5) == 15`.

**Подсказка:** запись `obj(x)` — это вызов `type(obj).__call__(obj, x)`; объект становится похож на функцию.


### 15. Наследование и super
Базовый класс `Animal(name)` с методом `speak()`. Класс `Dog(Animal)` добавляет поле `breed`, вызывает родительский `__init__` через `super()` и переопределяет `speak()`.

**Цель:** `isinstance(dog, Animal)` — `True`; `Dog("Rex", "husky").speak() == "Rex barks (husky)"`.

**Подсказка:** `super().__init__(name)` делегирует инициализацию родителю, чтобы не дублировать `self.name = name`.


### 16. Полиморфизм
Классы `Circle`, `Rectangle` и наследующий от него `Square` реализуют `area()`. Напишите функцию `total_area(figures)`, которая суммирует площади **без** `isinstance` и проверок типов.

**Цель:** `total_area([Circle(1), Rectangle(2, 3), Square(2)])` равна сумме трёх площадей.

**Подсказка:** duck typing: функцию не интересует тип — лишь бы у объекта был метод `area()`.


### 17. Ромбовидное наследование и MRO
Классы `A`, `B(A)`, `C(A)` и `D(B, C)`; в каждом определён метод `who()`. Выведите `D.__mro__` и определите, чья реализация `who()` вызовется у `D()`. Убедитесь, что `A` в цепочке ровно один раз.

**Цель:** имена в MRO: `D, B, C, A, object`; `D().who() == "B"`.

**Подсказка:** `super()` идёт не по списку родителей, а по линеаризации C3 — MRO.


### 18. Экономия памяти: __slots__
Создайте два одинаковых класса-точки: обычный и со `__slots__ = ("x", "y")`. Сравните наличие `__dict__` и `sys.getsizeof` у экземпляров. Проверьте, что экземпляру со слотами нельзя добавить новое поле.

**Цель:** `hasattr(slotted, "__dict__") == False`, `slotted.z = 3` бросает `AttributeError`; с учётом `__dict__` обычный экземпляр занимает заметно больше байт.

**Подсказка:** `__slots__` заменяет словарь атрибута фиксированным набором ячеек; атрибут вне списка добавить нельзя.


### 19. Фабрика на @classmethod
Класс `Product(name, price)`: `@staticmethod is_valid(data)` проверяет словарь, `@classmethod from_dict(cls, data)` собирает экземпляр (бросает `ValueError` на невалидных данных).

**Цель:** `Product.from_dict({"name": "milk", "price": 99})` — экземпляр Product; на `{"price": -5}` — `ValueError`.

**Подсказка:** в classmethod первым аргументом идёт сам класс `cls` — удобно для порождающих (factory) методов.


### 20. Сериализация объекта в JSON
Класс `User(name, age, tags)`: метод `to_json()` (через `json.dumps` с `default`) и `@classmethod from_json(cls, text)`. Добавьте `__eq__` сравнением полей.

**Цель:** `User.from_json(u.to_json()) == u` — круговое преобразование без потерь.

**Подсказка:** `json.dumps(obj, default=lambda o: o.__dict__)` сериализует любой простой объект; обратный ход — `cls(**json.loads(text))`.


## Сложные задачи


### 21. Дескриптор с валидацией
Реализуйте дескриптор `Positive`: при присваивании проверяет, что значение — положительное целое, иначе бросает `ValueError`; значение хранит в словаре экземпляра под служебным именем. Примените его в классе `Order` для полей `quantity` и `price`.

**Цель:** `o.quantity == 2`, а `o.quantity = -1` бросает `ValueError`.

**Подсказка:** дескриптор — класс с `__set_name__` (ловит имя поля), `__get__` и `__set__`; данные храните в `obj` под именем вида `_quantity`.


### 22. Авторегистрация через __init_subclass__
Базовый класс `Plugin` с пустым словарём `registry`. Реализуйте `__init_subclass__`, чтобы при объявлении `class Echo(Plugin, name="echo")` класс сам попадал в `Plugin.registry["echo"]`.

**Цель:** `Plugin.registry["echo"] is EchoPlugin` и `Plugin.registry["upper"]().run("hi") == "HI"`.

**Подсказка:** `__init_subclass__(cls, **kwargs)` — classmethod, который Python вызывает на родителе в момент объявления потомка; именованный аргумент `name` придёт прямо в него.


### 23. Registry и фабрика
Сделайте словарь-фабрику `FACTORY` и декоратор `@register`, который кладёт класс в фабрику по атрибуту `kind`. Реализуйте `create(kind, *args)` и две фигуры: `Circle` и `Square` с методом `area()`.

**Цель:** `create("circle", 2).area()` равна площади круга радиуса 2; неизвестный вид бросает `KeyError`.

**Подсказка:** декоратор — функция, принимающая класс, сохраняющая его в словарь и возвращающая его же без изменений.


### 24. Мини-ORM
Базовый класс `Model`: автоинкрементный `id`, `save()` в «таблицу» (список уровня класса), `all()`, `filter(**kw)` и `get(pk)`. Наследник `Note` не должен писать ни строчки кода.

**Цель:** после двух `save()` метод `Note.all()` отдаёт 2 строки, `Note.filter(done=False)` — 1, `Note.get(2)` — вторую заметку.

**Подсказка:** `type(self)._rows` — общий список для всех моделей; в `all()` отфильтруйте строки по `type(row) is cls`, чтобы таблицы наследников не смешивались.


### 25. Класс-матрица
Реализуйте `Matrix(data)` (список списков): `__getitem__`/`__setitem__` по паре `(i, j)`, сложение `+`, умножение матриц `*`, сравнение `==` и `__repr__`. Проверка согласованности размеров обязательна.

**Цель:** `Matrix([[1,2],[3,4]]) * Matrix([[5,6],[7,8]]) == Matrix([[19,22],[43,50]])`, сложение и доступ по индексу работают.

**Подсказка:** храните данные как список списков; для умножения элемент `(i, j)` — скалярное произведение строки `i` первой и столбца `j` второй.


### 26. Очередь с полным протоколом
Класс `Queue` поверх `collections.deque`: `put`, `get` (FIFO, с пустой — `IndexError`), `__len__`, `__bool__`, `__contains__` и `__iter__`, который НЕ потребляет элементы.

**Цель:** `list(q) == [1, 2, 3]` при этом длина очереди не изменилась; `q.get()` отдаёт старейший элемент.

**Подсказка:** `popleft()` — снятие с головы очереди; для безопасной итерации возвращайте `iter(list(self._dq))`.


### 27. Паттерн Observer на классах
Класс `EventEmitter`: `on(event, handler)` подписка, `off(event, handler)` отписка, `emit(event, **data)` — вызывает всех подписчиков с словарём данных.

**Цель:** после `emit` обработчик получил `{"x": 1, "y": 2}`; после `off` новые события игнорируются.

**Подсказка:** словарь `event -> список коллбэков`; `setdefault` экономит проверку наличия ключа.


### 28. Дробь: арифметика и нормализация
Класс `Fraction(num, den)`: при создании сокращение через `math.gcd`, знаменатель всегда положителен, ноль в знаменателе — `ZeroDivisionError`. Реализуйте `+`, `*`, `__eq__`, `__repr__` (целые выводятся без знаменателя).

**Цель:** `Fraction(1, 2) + Fraction(1, 6) == Fraction(2, 3)` и `Fraction(2, 4) == Fraction(1, 2)`.

**Подсказка:** приводите к общему знаменателю и сокращайте результат в конструкторе — тогда равенство проверяется попарно.


### 29. Money: сравнение через total_ordering
Класс `Money` хранит копейки. Реализуйте `+`, `__eq__`, `__lt__`, декоратор `functools.total_ordering`, `__str__` вида `20.00 rub` и `@classmethod from_rubles`.

**Цель:** `sorted` по списку Money работает, `Money(150) < Money.from_rubles(2)`, `str` даёт `20.00 rub`.

**Подсказка:** `total_ordering` выведет `<=`, `>`, `>=` из ваших `__eq__` и `__lt__`.


### 30. Вложенная сериализация + фабрика
Класс `Order(customer)` со списком объектов `Position(title, qty, price)`. Реализуйте `to_dict`/`from_dict` у обоих классов, `to_json()` и `total()`. `from_dict` обязан восстановить вложенные `Position`, а не словари.

**Цель:** `Order.from_dict(json.loads(o.to_json())).to_dict() == o.to_dict()`, и каждый элемент `positions` — экземпляр `Position`.

**Подсказка:** сериализация рекурсивна: `to_dict` порядка вызывает `to_dict` каждой позиции; `from_dict` собирает вложенные объекты через их `from_dict`.
